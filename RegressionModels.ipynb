```python
# ============================================================
# STEP 1: IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import PolynomialFeatures

from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# STEP 2: EVALUATION FUNCTION
# ============================================================

def evaluate(model_name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    return {
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    }


# ============================================================
# STEP 3: CREATE RESULTS LIST
# ============================================================

results = []


# ============================================================
# STEP 4: BASELINE MODEL
# Predicts the mean value of y_train
# ============================================================

baseline_predictions = np.full(
    len(y_test),
    y_train.mean()
)

results.append(
    evaluate(
        "Mean Baseline",
        y_test,
        baseline_predictions
    )
)

print("Baseline predictions:")
print(baseline_predictions)


# ============================================================
# STEP 5: CHECK FEATURE TYPES
# ============================================================

# If x_train is a DataFrame, find numerical and categorical columns

numerical_features = x_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = x_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# STEP 6: PREPROCESSOR
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numbers",
            SimpleImputer(strategy="median"),
            numerical_features
        ),

        (
            "category",
            Pipeline(
                steps=[
                    (
                        "imputer",
                        SimpleImputer(strategy="most_frequent")
                    ),

                    (
                        "encoder",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            drop="first"
                        )
                    )
                ]
            ),
            categorical_features
        )
    ]
)

print("\nPreprocessor created successfully.")


# ============================================================
# STEP 7: MULTIPLE LINEAR REGRESSION
# ============================================================

multiple_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

# Train model
multiple_model.fit(x_train, y_train)

# Make predictions
multiple_prediction = multiple_model.predict(x_test)

# Evaluate
results.append(
    evaluate(
        "Multiple Regression",
        y_test,
        multiple_prediction
    )
)

print("\nMultiple Linear Regression completed.")


# ============================================================
# STEP 8: POLYNOMIAL REGRESSION
# ============================================================

polynomial_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "polynomial",
            PolynomialFeatures(
                include_bias=False
            )
        ),

        (
            "model",
            LinearRegression()
        )
    ]
)

print("\nPolynomial model created.")


# ============================================================
# STEP 9: SELECT BEST POLYNOMIAL DEGREE
# ============================================================

degree_search = GridSearchCV(
    polynomial_model,

    param_grid={
        "polynomial__degree": [1, 2, 3, 4]
    },

    cv=5,

    scoring="neg_mean_squared_error"
)


# Train GridSearchCV
degree_search.fit(
    x_train,
    y_train
)


# Display best degree
print(
    "\nBest polynomial degree:",
    degree_search.best_params_
)


# ============================================================
# STEP 10: POLYNOMIAL PREDICTIONS
# ============================================================

polynomial_predictions = degree_search.predict(
    x_test
)

print("\nPolynomial predictions:")
print(polynomial_predictions)


# ============================================================
# STEP 11: EVALUATE POLYNOMIAL REGRESSION
# ============================================================

results.append(
    evaluate(
        "Polynomial Regression",
        y_test,
        polynomial_predictions
    )
)


# ============================================================
# STEP 12: DISPLAY ALL RESULTS
# ============================================================

results_df = pd.DataFrame(results)

print("\n================ MODEL RESULTS ================\n")

print(results_df)
```
