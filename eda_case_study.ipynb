```python
#Understanding the missing values and filling the missing values
data.isnull().sum()
```

```python
data.isnull().sum()
```

```python
#we can fill the Agecolumn missing values with mean
data['Age'] = data['Age'].fillna(data['Age'].mean(),inplace=True)
data.isnull().sum()
```

```python
data['Embarked'] = data['Embarked'].fillna('S')
```

```python
data['Age'] = data['Age'].fillna(data['Age'].mean())
data.isnull().sum()
```

```python
data.drop(columns=['Cabin'],inplace=True)
```

```python
data.isnull().sum()
```

```python
#Feture Engineering -->Feature Engineering (It is the process of converting
#categorical values into numerical)
#In above case we have Sex,Pclass,Embarked
data.columns
```

```python
#get _dummies()
pclass = pd.get_dummies(data['Pclass'])
pclass
```

```python
#get _dummies()
sex = pd.get_dummies(data['Sex'],drop_first = True,dtype=int)
sex
```

```python
#get _dummies()
embarked= pd.get_dummies(data['Embarked'],drop_first = True,dtype=int)
embarked
```

```python
gender = pd.get_dummies(data['Sex'],drop_first=True,dtype=int)
gender
```

```python
#Concatenation of all tables
final_data = pd.concat([data,pclass,gender,embarked],axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Pclass','Name','Sex','Ticket','Embarked'],inplace=True)
final_data.head()
```
